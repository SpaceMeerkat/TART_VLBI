# Check whether inter-TART acquisitions overlap in time

```python
import numpy as np
import sys
import os

# Get parent directory
parent_dir = os.path.abspath("..")

# Add it to sys.path
if parent_dir not in sys.path:
    sys.path.insert(0, parent_dir)
    
from pipeline.dataloader import raw_data_loader
```

```python
TARTS = {
    'rhodes': 'za-rhodes',
    'namibia': 'na-unam',
}

DATASETS = {
    'rhodes': 'rhodes/data_2026-02-13_14_39_25.231047.hdf',
    'namibia': 'namibia/data_2026-02-13_14_39_25.193869.hdf'
}

loader = raw_data_loader("/home/jdawson/repos/TART/datasets")
data_rhodes, time_rhodes, Fs_rhodes = loader.load_preprocessed_data(DATASETS['rhodes'])
data_namibia, time_namibia, Fs_namibia = loader.load_preprocessed_data(DATASETS['namibia'])
Fs_baseband = data_rhodes['Fs']

t_rhodes_interp = np.linspace(time_rhodes[0], time_rhodes[-1], len(time_rhodes))
t_namibia_interp = np.linspace(time_namibia[0], time_namibia[-1], len(time_namibia))
time_window = loader.get_overlapping_time_window([t_rhodes_interp, t_namibia_interp])

print(f"\nDatasets have an overlapping observation window from: {time_window[0]} -> {time_window[1]} unix.")
print(f"Overlapping length: {time_window[1] - time_window[0]} seconds.")
```

## Take the half way point as the observation time for getting GPS locations

```python
obs_time = np.sum(time_window)/2
print(f"Chosen observation time is: {obs_time}")
```
